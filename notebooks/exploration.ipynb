```python
import pandas as pd
import numpy as np
import pyxlsb
import matplotlib.pyplot as plt
```

```python
data=pd.read_excel('../data/input/20210614 Ecommerce sales.xlsb')
data.head()
```

```python
print("Nombre de lignes :", len(data))
print("Nombre de natures uniques :", data["Nature"].nunique())
print("\nNatures :", data["Nature"].unique())
print("\nRépartition :")
print(data["Nature"].value_counts())
```

```python
top5 = data["Nature"].value_counts().head(5)

top5.plot(kind="barh")
plt.title("Top 5 des natures les plus présentes")
plt.xlabel("Nombre d'occurrences")
plt.ylabel("Nature")
plt.gca().invert_yaxis()
plt.show()
```
